```python
#guess nymber game

import random

def check_the_number(guess,final_number):
    if guess == final_number:
        return False
    elif guess > final_number:
        print("too high.\nguess again")
    else:
        print("too low.\nguess again")
        
EASY_LEVEL = 10
HIGH_LEVEL = 5        

def set_dificulty(dificulty_level):
    if dificulty_level =="easy":
       return EASY_LEVEL
    elif dificulty_level =="hard":
       return HIGH_LEVEL
     
def numbre_guessing():
    turns = 0
    print("Welcome to the Number Guessing Game!")
    print("I'm thinking of a number between 1 and 100.")
    final_number = random.randint(0,100)
    dificulty_level = input("Choose a difficulty. Type 'easy' or 'hard': ").lower()
    turns = set_dificulty(dificulty_level)
    guess = 0
    while guess !=final_number :
        if  turns > 0:
            guess = int(input("make a guess:"))
            print("make a guess:",guess)
            result = check_the_number(guess,final_number)
            if result ==False:
                break  
            else:
                turns += -1
                print(f"You have {turns} attempts remaining to guess the number.")
        else: 
            print("You've run out of guesses, you lose.")
            return
    print (f"you got it, the answer was {final_number}")    
    
    
    
numbre_guessing()
    
```
