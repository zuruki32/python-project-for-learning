```python
import sys
import io
import unittest

# تابع خودت (همونی که دادی، فقط یه کم ادیت کردم تا شبیه تست بشه)
def make_bread():
    x = "minutes."
    y = "and"
    z = "of"
    f  = "hours."
    h = "for"
    j = "at"
    a = "in"
    m = "the"
    n = "g"
    b = "ml"
    for i in range(0,6):
        if i == 1:
            print(f"{i}. Mix 500{n} {z} Flour, 10{n} Yeast {y} 300{b} Water {a} a bowl.")
        elif i==2:
            print(f"{i}. Knead {m} dough {h} 10 {x}")
        elif i==3:
            print(f"{i}. Add 3{n} {z} Salt.")
        elif i==4:
            print(f"{i}. Leave to rise {h} 2 {f}")
        elif i==5:
            print(f"{i}. Bake {j} 200 degrees C {h} 30 {x}")

# کلاس تست
class TestBread(unittest.TestCase):
    def test_output(self):
      
        captured = io.StringIO()
        old_stdout = sys.stdout
        sys.stdout = captured
        
        make_bread()  
        
        sys.stdout = old_stdout
        output = captured.getvalue()
        
        expected = "1. Mix 500g of Flour, 10g Yeast and 300ml Water in a bowl.\n2. Knead the dough for 10 minutes.\n3. Add 3g of Salt.\n4. Leave to rise for 2 hours.\n5. Bake at 200 degrees C for 30 minutes.\n"
        
   
        self.assertEqual(expected, output, "خروجی با انتظار فرق داره")


unittest.main(argv=[''], exit=False)
```

```python
#bmi calculator

def bmi_calculate():
    height = float(input("please enter your height in m"))
    weight = float(input("please enter your weight in kg"))
    bmi = round(weight/(height**2))
    if bmi < 18.5:
        print(f"Your BMI is {bmi}, you are underweight.")
    elif 25 > bmi > 18.5:
        print(f"Your BMI is {bmi}, you have a normal weight.")
    elif 30 > bmi > 25:
        print(f"Your BMI is {bmi}, you are slightly overweight.")
    elif 35 > bmi > 30:
        print(f"Your BMI is {bmi}, you are obese.")
    elif bmi > 35:
        print(f"Your BMI is {bmi}, you are clinically obese.")
        
        
bmi_calculate()
```

```python
#leaf year
#if yeaer can devide to 4 and shouldnt devide to 100 but if it did then it should devided to  400 too

def leap_year(year):
    is_leap_year =True
    if year%4 == 0:
        if year%100 !=0:
            return is_leap_year
        else:
            if year%400 ==0:
                return is_leap_year
            else:
                is_leap_year =False
                return is_leap_year
    else:
        is_leap_year =False
        return is_leap_year

leap_year(1989)
```

```python
#pizza order app
def pizza_app():
    print("welcome to python pizza deliveris/n")
    bill =0
    size = input("what size of pizza do you want? S,M or L?/n")
    peperoni = input("do you want peperoni on your pizza? Y or N?")
    extra_cheese = input("do you want extra cheese? Y or N?")
    if size == "S":
        bill +=15
        if peperoni == "Y":
            bill +=2
    elif size == "M":
        bill +=20
        if peperoni == "Y":
            bill +=3
    elif size =="L":
        bill +=25
        if peperoni == "Y":
            bill +=3
    else:
        print("you typed a wrong input")
    if extra_cheese== "Y":
        bill +=1
    print(f"Your final bill is: ${bill}.")
pizza_app()    
```

```python
#love calculator
#2 new function here first one is string.lower() and the other is string.count(character)that would check how many times a character happend in string
def love_calculater():
    name1 = input("please write the first name.")
    name2 = input("please write the second name.")
    string1 = "true"
    string2 = "love"
    score1 = 0
    score2 = 0
    total = name1 + name2
    make_lowercase = total.lower()
    for i in string1:
        score1 += make_lowercase.count(i)
    for j in  string2:
        score2 +=make_lowercase.count(j)
    final_score =int(str(score1) + str(score2))
    
    if final_score<10 or final_score>90:
        print(f"Your score is {final_score}, you go together like coke and mentos.")
    elif 40<final_score<50:
        print(f"Your score is {final_score}, you are alright together.")
    else:
        print(f"Your score is {final_score}.")

love_calculater()
    
    
            
```

```python
#calculator
#we can also in additon of variable place func inside of the value of dic
def add(x,y):
    return x+y
def minus(x,y):
    return x-y
def multi(x,y):
    return x*y
def devide(x,y):
    return x/y

operator_dic = {"+":add,"-":minus,"*":multi,"/":devide}

def take_result(first_number,operator,second_number):
    white_board = ""
    result = 0
    if second_number == 0 and operator =="/":
        print("the 0 cant be use in deviding please try another number")
        return 
    if operator in operator_dic:
        result = operator_dic[operator](first_number,second_number)
    white_board = str(first_number) + operator+ str(second_number) + "=" + str(result)
    print(white_board) 
    return result
    
def calculator():
    calculator_on = True
    previous_result_use = False
    previous_result = 0
    while calculator_on :
        if previous_result_use == False:
            first_number = float(input("whats the first number?"))
            operator = input("+\n-\n*\n/\npick an operator:")
            second_number = float(input("whats the second number?"))
            result_final = take_result(first_number,operator,second_number)
            if result_final == None:
               return
            else: 
               print(result_final)
            previous_result =result_final
            countinue = input(f"type 'y' to continue or with {result_final} or type 'n' to start a new calculation").lower()
            if countinue == "y":
                previous_result_use = True
        elif previous_result_use==True:
            first_number = previous_result
            print(first_number)
            operator = input("+\n-\n*\n/\npick an operator:")
            second_number = int(input("whats the second number?"))
            result_final = take_result(first_number, operator,second_number)
            previous_result =result_final
            countinue = input(f"type 'y' to continue or with {result_final} or type 'n' to start a new calculation").lower()
            if countinue == "y":
                previous_result_use = True
            elif countinue == "n":
                previous_result_use = False
            
        
calculator()       
            
```
